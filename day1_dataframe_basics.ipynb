```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Day1").getOrCreate()
```

```python
#create sample data

data = [ (1, "Divya", 28, 50000),
         (2, "Keerthi", 27, 40000),
         (3, "Uday", 34, 100000),
         (4, "Kumar", 28, 50000) ]

columns = ["ID", "Name", "Age", "Salary"]

df = spark.createDataFrame(data, columns)

df.show()
```

```python
#basic operations
#1. select columns

df.select("ID", "Name").show()
```

```python
#2. filter Data

df.filter(df.Salary > 50000).show()
```

```python
#3. Add new column

from pyspark.sql.functions import col

df = df.withColumn("Salary_in_n", col("Salary")/1000)
df.show()
```

```python
df.show()
```

```python
#4. Rename Column

df = df.withColumnRenamed("Name", "Employee_name")
df.show()
```

```python
#5 drop column

df = df.drop("Salary_in_n")
df.show()
```

```python
#6 pandas for better dispaly

df.toPandas()
```

```python
#7 left alligned result (cutting off the spaces)

df.show(truncate=False)
```

```python
#8 pandas conversion (every col converts into str)

pdf = df.toPandas().astype(str)
pdf
```

```python
#9  left indent using pandas

df.toPandas().style.set_properties(**{'text-align' : 'left'})
```

```python
#minitasks
#1. Filter employees with salary > 55000
#2. Add bonus column = salary * 0.1
#3. Show only name + bonus

from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("minitasks_of_day1").getOrCreate()

data = [(1, "Divya", 23, 60000), (2, "Keerthi", 25, 50000), (3, "Uday", 28, 100000), (4, "Kumar", 30, 55000) ]
columns = ["ID", "Name", "Age", "Salary"]

df = spark.createDataFrame(data, columns)

df = df.filter(df.Salary >55000)
df.show(truncate=False)

df = df.withColumn("Bonus", col("Salary") * 0.1)
df.show(truncate=False)

df = df.select("Name", "Bonus")
df.show(truncate=False)


```

```python
#10 adding new column

from pyspark.sql import SparkSession
from pyspark.sql.functions import lit

spark = SparkSession.builder.appName("minitasks_of_day1").getOrCreate()

data = [(1, "Divya", 23, 60000), (2, "Keerthi", 25, 50000), (3, "Uday", 28, 100000), (4, "Kumar", 30, 55000) ]
columns = ["ID", "Name", "Age", "Salary"]

df = spark.createDataFrame(data, columns)

df = df.withColumn("Result", lit("Pass"))

df.show(truncate=False)
```

```python
#11 based on condition

from pyspark.sql.functions import when, col

df = df.withColumn("Result", when(col("Salary") > 50000, "High").otherwise("Low"))

df.show(truncate=False)
df.count()

df.collect()

```
