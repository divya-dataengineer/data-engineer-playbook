```python
#1. Transformations
#Transformations are lazy operations
#They do NOT execute immediately
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Transformation_Actions_LazyEvaluation").getOrCreate()

data = [(1, "Divya", 28, 50000),
       (2, "Keerthi", 27, 40000),
       (3, "Uday", 28, 100000),
       (4, "Kumar", 31, 30000) ]

columns = ["id", "name", "age", "salary"]

df = spark.createDataFrame(data, columns)
df.show(truncate=False)
```

```python
df_filtered = df.filter(df.salary > 50000)

df_selected = df.select(df.age > 27)

#they donot give result but executed successfully.
#👉 No output yet
#👉 Spark is just planning
```

```python
#2. Actions
#Now Spark actually runs the job ✔

df.show(truncate=False)

df.count()

df.collect()

df_filtered.show(truncate=False)

df_selected.show(truncate=False)

```

```python
#3. Lazy Evaluation (MOST IMPORTANT)
#Spark waits until action to execute everything

df1 = df.filter(df.salary > 50000)
df2 = df1.select("name")

df2.show(truncate=False)

#Step 1 → Not executed
#Step 2 → Not executed
#Step 3 → Everything runs together ✔

```

```python
#📦 Behind the Scenes

#👉 Spark builds a DAG (Directed Acyclic Graph)

#Then:

#Optimizes it
#Executes efficiently

#🎯 Why Lazy Evaluation?

#👉 Performance optimization:

#Avoid unnecessary steps
#Combine operations
#Reduce data movement
```
